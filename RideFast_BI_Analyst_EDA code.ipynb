```python
rides['request_time'] = pd.to_datetime(rides['request_time'])
rides['pickup_time'] = pd.to_datetime(rides['pickup_time'])
rides['dropoff_time'] = pd.to_datetime(rides['dropoff_time'])

users['signup_date'] = pd.to_datetime(users['signup_date'])
users['last_ride_date'] = pd.to_datetime(users['last_ride_date'])

drivers['signup_date'] = pd.to_datetime(drivers['signup_date'])

tickets['created_at'] = pd.to_datetime(tickets['created_at'])
```

```python
rides.info()
```

```python
rides['request_date'] = rides['request_time'].dt.date
rides['request_month'] = rides['request_time'].dt.to_period('M').astype(str)
rides['request_hour'] = rides['request_time'].dt.hour
rides['request_day'] = rides['request_time'].dt.day_name()

rides['is_completed'] = rides['status'] == 'completed'
rides['is_cancelled'] = rides['status'].isin(['cancelled_driver', 'cancelled_user'])

rides['wait_time_minutes'] = (
    (rides['pickup_time'] - rides['request_time']).dt.total_seconds() / 60
)
```

Core KPI summary

```python
total_requests = len(rides)
completed_rides = rides['is_completed'].sum()
completion_rate = (completed_rides / total_requests) * 100
total_revenue = rides['fare_amount'].sum()
avg_fare = rides[rides['is_completed']]['fare_amount'].mean()

print("Total Requests:", total_requests)
print("Completed Rides:", completed_rides)
print("Completion Rate:", round(completion_rate, 2), "%")
print("Total Revenue:", round(total_revenue, 2))
print("Average Completed Ride Fare:", round(avg_fare, 2))
```

Promo usage analysis

```python
rides['promo_code_used'].value_counts()
```

```python
(rides['promo_code_used'].value_counts(normalize=True) * 100).round(2)
```

Promo effectiveness analysis

```python
promo_analysis = rides.groupby('promo_code_used').agg(
    total_requests=('ride_id', 'count'),
    completed_rides=('is_completed', 'sum'),
    total_revenue=('fare_amount', 'sum'),
    avg_fare=('fare_amount', 'mean'),
    avg_discount=('promo_discount', 'mean')
)

promo_analysis['completion_rate'] = (
    promo_analysis['completed_rides'] / promo_analysis['total_requests'] * 100
)

promo_analysis.round(2)
```

City performance

```python
city_perf = rides.groupby('city').agg(
    total_requests=('ride_id', 'count'),
    completed_rides=('is_completed', 'sum'),
    total_revenue=('fare_amount', 'sum')
)

city_perf['completion_rate'] = (
    city_perf['completed_rides'] / city_perf['total_requests'] * 100
)

city_perf['cancellation_rate'] = 100 - city_perf['completion_rate']

city_perf.sort_values('cancellation_rate', ascending=False).round(2)
```

Status breakdown by city

```python
city_status = pd.crosstab(
    rides['city'],
    rides['status'],
    normalize='index'
) * 100

city_status.round(2).sort_values('cancelled_driver', ascending=False)
```

Vehicle type performance

```python
vehicle_perf = rides.groupby('vehicle_type').agg(
    total_requests=('ride_id', 'count'),
    completed_rides=('is_completed', 'sum'),
    avg_wait=('wait_time_minutes', 'mean'),
    avg_fare=('fare_amount', 'mean')
)

vehicle_perf['completion_rate'] = (
    vehicle_perf['completed_rides'] / vehicle_perf['total_requests'] * 100
)

vehicle_perf.round(2)
```

Rides by hour

```python
hourly_perf = rides.groupby('request_hour').agg(
    total_requests=('ride_id', 'count'),
    completed_rides=('is_completed', 'sum'),
    avg_wait=('wait_time_minutes', 'mean')
)

hourly_perf['completion_rate'] = (
    hourly_perf['completed_rides'] / hourly_perf['total_requests'] * 100
)

hourly_perf.round(2)
```
